```python
plt.figure(figsize=(10, 6))

plt.hist(titanic['Age'].dropna(), bins=20, edgecolor='black', alpha=0.7, color='skyblue')

plt.title('Titanic: Распределение возраста пассажиров', fontsize=14)
plt.xlabel('Возраст', fontsize=12)
plt.ylabel('Количество пассажиров', fontsize=12)

if not titanic['Age'].isna().all():
    mean_age = titanic['Age'].mean()
    plt.axvline(mean_age, color='red', linestyle='--', linewidth=2,
                label=f'Средний возраст: {mean_age:.1f}')
    plt.legend(fontsize=12)

plt.grid(alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

colors = {'Iris-setosa': 'red', 'Iris-versicolor': 'green', 'Iris-virginica': 'blue'}

for species in iris['Species'].unique():
    subset = iris[iris['Species'] == species]
    plt.scatter(subset['PetalLengthCm'], subset['PetalWidthCm'],
               label=species, alpha=0.7, s=50,
               color=colors.get(species, 'gray'))

plt.title('Iris: Длина vs Ширина лепестка', fontsize=14)
plt.xlabel('Длина лепестка (см)', fontsize=12)
plt.ylabel('Ширина лепестка (см)', fontsize=12)

plt.legend(fontsize=12)

plt.grid(alpha=0.3)

plt.show()
```
